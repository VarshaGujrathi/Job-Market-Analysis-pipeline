```python
# RAW DATA INSPECTION - GREENHOUSE

GREENHOUSE_RAW_PATH = (
    "s3://job-market-analysis-pipeline/"
    "raw-data/greenhouse/"
    "ingestion_date=2026-10-06/*.json"
)

greenhouse_raw = spark.read.json(
    GREENHOUSE_RAW_PATH
)

print("Greenhouse RAW schema:")
greenhouse_raw.printSchema()

print("\nNumber of records:")
print(greenhouse_raw.count())

display(greenhouse_raw.limit(5))
```

```python
# EXPLODE GREENHOUSE JOBS

greenhouse_jobs = greenhouse_raw.selectExpr(
    "explode(jobs) AS job"
)

print("Greenhouse job schema:")
greenhouse_jobs.select("job.*").printSchema()

display(
    greenhouse_jobs.select("job.*").limit(5)
)
```

```python
greenhouse_job_df = greenhouse_jobs.select("job.*")

print("Greenhouse columns:")
print(greenhouse_job_df.columns)
```

```python
# RAW DATA INSPECTION - LEVER

LEVER_RAW_PATH = (
    "s3://job-market-analysis-pipeline/"
    "raw-data/lever/"
    "ingestion_date=2026-10-06/*.json"
)

lever_raw = spark.read.json(
    LEVER_RAW_PATH
)

print("Lever RAW schema:")
lever_raw.printSchema()

print("\nNumber of records:")
print(lever_raw.count())

display(
    lever_raw.limit(5)
)
```

```python
# EXPLODE LEVER JOBS


lever_jobs = lever_raw.selectExpr(
    "explode(jobs) AS job"
)

print("Lever job schema:")

lever_jobs.select(
    "job.*"
).printSchema()

display(
    lever_jobs.select(
        "job.*"
    ).limit(5)
)
```

```python
lever_job_df = lever_jobs.select("job.*")

print("Lever columns:")
print(lever_job_df.columns)
```

```python
print("==========================================")
print("JOB COUNTS")
print("==========================================")

print(
    "Greenhouse jobs:",
    greenhouse_job_df.count()
)

print(
    "Lever jobs:",
    lever_job_df.count()
)
```

```python
from pyspark.sql.functions import col, sum

print("Greenhouse NULL counts:")

display(
    greenhouse_job_df.select(
        [
            sum(
                col(c).isNull().cast("int")
            ).alias(c)
            for c in greenhouse_job_df.columns
        ]
    )
)
```

```python
print("Lever NULL counts:")

display(
    lever_job_df.select(
        [
            sum(
                col(c).isNull().cast("int")
            ).alias(c)
            for c in lever_job_df.columns
        ]
    )
)
```
