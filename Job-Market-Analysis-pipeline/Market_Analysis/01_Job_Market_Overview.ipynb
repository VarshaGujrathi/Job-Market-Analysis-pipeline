```sql
-- Which companies have the highest hiring demand?

SELECT
    company,
    COUNT(*) AS job_count
FROM job_market.silver.jobs_clean
GROUP BY company
ORDER BY job_count DESC;
```

```sql
-- Top 10 job titles are most in demand?

SELECT
    job_title,
    job_count,
    company_count,
    demand_percentage
FROM job_market.gold.job_demand
ORDER BY job_count DESC
LIMIT 10;
```

```sql
-- Top 10 job titles are offered by the largest number of companies?

SELECT
    job_title,
    company_count,
    job_count
FROM job_market.gold.job_demand
ORDER BY company_count DESC
LIMIT 10;
```

```sql
-- What is the overall distribution of remote, hybrid and onsite jobs?

SELECT
    CASE
        WHEN is_remote = TRUE THEN 'Remote'
        WHEN LOWER(workplace_type) = 'hybrid' THEN 'Hybrid'
        WHEN LOWER(workplace_type) = 'onsite' THEN 'Onsite'
        ELSE 'Not Specified'
    END AS workplace_model,

    COUNT(*) AS job_count

FROM job_market.silver.jobs_clean

GROUP BY
    CASE
        WHEN is_remote = TRUE THEN 'Remote'
        WHEN LOWER(workplace_type) = 'hybrid' THEN 'Hybrid'
        WHEN LOWER(workplace_type) = 'onsite' THEN 'Onsite'
        ELSE 'Not Specified'
    END

ORDER BY job_count DESC;
```

```sql
-- Which companies have the highest proportion of remote jobs?

SELECT
    company,
    COUNT(*) AS total_jobs,

    SUM(
        CASE
            WHEN is_remote = TRUE THEN 1
            ELSE 0
        END
    ) AS remote_jobs,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN is_remote = TRUE THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS remote_percentage

FROM job_market.silver.jobs_clean

GROUP BY company

ORDER BY remote_percentage DESC;
```
