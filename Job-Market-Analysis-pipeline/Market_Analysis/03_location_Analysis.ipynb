```sql
-- Which locations have the highest number of job openings?

SELECT
    location,
    job_count,
    company_count,
    job_title_count,
    demand_percentage
FROM job_market.gold.location_demand
ORDER BY job_count DESC
LIMIT 10;
```

```sql
-- Which locations are being targeted by the largest number of companies?

SELECT
    location,
    company_count,
    job_count,
    job_title_count
FROM job_market.gold.location_demand
ORDER BY company_count DESC, job_count DESC
LIMIT 10;
```

```sql
-- Which locations have the highest proportion of remote jobs?

SELECT
    location,
    job_count,
    remote_job_count,
    remote_percentage,
    company_count
FROM job_market.gold.location_demand
WHERE job_count >= 10
ORDER BY remote_percentage DESC
LIMIT 10;
```
