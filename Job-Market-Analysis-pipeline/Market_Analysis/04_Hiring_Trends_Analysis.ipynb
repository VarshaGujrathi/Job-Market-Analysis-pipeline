```sql
-- How does overall job-posting volume change month by month?
SELECT
    posting_month,
    job_count,
    company_count,
    job_title_count
FROM job_market.gold.hiring_trends
ORDER BY posting_month;
```

```sql
-- How much has hiring changed compared with the previous month?

SELECT
    posting_month,
    job_count,
    previous_month_job_count,
    mom_job_change,
    mom_growth_percentage,
    hiring_trend
FROM job_market.gold.hiring_trends
ORDER BY posting_month;
```

```sql
-- Which months performed above or below the overall average monthly hiring volume?
SELECT
    posting_month,
    job_count,
    ROUND(AVG(job_count) OVER (), 2) AS average_monthly_jobs,
    CASE
        WHEN job_count > AVG(job_count) OVER () THEN 'Above Average'
        WHEN job_count < AVG(job_count) OVER () THEN 'Below Average'
        ELSE 'Average'
    END AS hiring_level
FROM job_market.gold.hiring_trends
ORDER BY posting_month DESC;
```

```sql
-- How is remote hiring changing relative to total hiring?

SELECT
    posting_month,
    job_count,
    remote_job_count,
    remote_percentage
FROM job_market.gold.hiring_trends
ORDER BY posting_month;
```
