```sql
-- What technical skills are most frequently required across job postings?

SELECT
    skill,
    job_count,
    demand_percentage,
    company_count,
    location_count
FROM job_market.gold.skill_demand
ORDER BY job_count DESC
LIMIT 10;
```

```sql
-- Which technical skills are most frequently required in remote job opportunities?

SELECT
    skill,
    remote_job_count,
    remote_percentage,
    job_count
FROM job_market.gold.skill_demand
ORDER BY remote_job_count DESC
LIMIT 10;
```

```sql
-- Which companies are hiring most for specific technical skills?

SELECT
    company,
    skill,
    job_count
FROM job_market.gold.company_skill_demand
WHERE LOWER(skill) IN (
    'python',
    'sql',
    'pyspark',
    'databricks',
    'aws'
)
ORDER BY skill,
         job_count DESC;
```

```sql
-- Which companies have the broadest range of technical skill requirements?

SELECT
    company,
    COUNT(DISTINCT skill) AS number_of_skills
FROM job_market.gold.company_skill_demand
GROUP BY company
ORDER BY number_of_skills DESC
LIMIT 10;
```
